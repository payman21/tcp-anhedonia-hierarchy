## BAPQ: Broad Autism Phenotype Questionnaire

```python
bapq_imp = pd.read_csv('../phenotype/imputed/imputed_bapq01.csv')
```

```python
bapq_imp.columns
```

```python
# List of all BAPQ questions
bapq_questions = ['bapq_q1', 'bapq_q2', 'bapq_q3', 'bapq_q4', 'bapq_q5',
       'bapq_q6', 'bapq_q7', 'bapq_q8', 'bapq_q9', 'bapq_q10', 'bapq_q11',
       'bapq_q12', 'bapq_q13', 'bapq_q14', 'bapq_q15', 'bapq_q16', 'bapq_q17',
       'bapq_q18', 'bapq_q19', 'bapq_q20', 'bapq_q21', 'bapq_q22', 'bapq_q23',
       'bapq_q24', 'bapq_q25', 'bapq_q26', 'bapq_q27', 'bapq_q28', 'bapq_q29',
       'bapq_q30', 'bapq_q31', 'bapq_q32', 'bapq_q33', 'bapq_q34', 'bapq_q35',
       'bapq_q36']

# Index mapping function (adjusting from 1-based to 0-based index)
def get_bapq_questions(indices):
    return [bapq_questions[i - 1] for i in indices]

# BAPQ Subscales
aloof = get_bapq_questions([1, 5, 9, 12, 16, 18, 23, 25, 27, 28, 31, 36])
pragmatic_language = get_bapq_questions([2, 4, 7, 10, 11, 14, 17, 20, 21, 29, 32, 34])
rigid = get_bapq_questions([3, 6, 8, 13, 15, 19, 22, 24, 26, 30, 33, 35])

# Print to verify
print("Aloof:", aloof)
print("Pragmatic Language:", pragmatic_language)
print("Rigid:", rigid)
```

```python
# Define column groups
aloof_cols = ['bapq_q1', 'bapq_q5', 'bapq_q9', 'bapq_q12', 'bapq_q16', 'bapq_q18', 
              'bapq_q23', 'bapq_q25', 'bapq_q27', 'bapq_q28', 'bapq_q31', 'bapq_q36']

prag_lang_cols = ['bapq_q2', 'bapq_q4', 'bapq_q7', 'bapq_q10', 'bapq_q11', 'bapq_q14', 
                  'bapq_q17', 'bapq_q20', 'bapq_q21', 'bapq_q29', 'bapq_q32', 'bapq_q34']

rigid_cols = ['bapq_q3', 'bapq_q6', 'bapq_q8', 'bapq_q13', 'bapq_q15', 'bapq_q19', 
              'bapq_q22', 'bapq_q24', 'bapq_q26', 'bapq_q30', 'bapq_q33', 'bapq_q35']

all_bapq_cols = ['bapq_q1', 'bapq_q2', 'bapq_q3', 'bapq_q4', 'bapq_q5',
                 'bapq_q6', 'bapq_q7', 'bapq_q8', 'bapq_q9', 'bapq_q10', 'bapq_q11',
                 'bapq_q12', 'bapq_q13', 'bapq_q14', 'bapq_q15', 'bapq_q16', 'bapq_q17',
                 'bapq_q18', 'bapq_q19', 'bapq_q20', 'bapq_q21', 'bapq_q22', 'bapq_q23',
                 'bapq_q24', 'bapq_q25', 'bapq_q26', 'bapq_q27', 'bapq_q28', 'bapq_q29',
                 'bapq_q30', 'bapq_q31', 'bapq_q32', 'bapq_q33', 'bapq_q34', 'bapq_q35',
                 'bapq_q36']

# Compute new columns
bapq_imp['bapq_aloof_total'] = bapq_imp[aloof_cols].sum(axis=1)
bapq_imp['bapq_aloof_ave'] = bapq_imp[aloof_cols].mean(axis=1)

bapq_imp['bapq_prag_lang_total'] = bapq_imp[prag_lang_cols].sum(axis=1)
bapq_imp['bapq_prag_lang_ave'] = bapq_imp[prag_lang_cols].mean(axis=1)

bapq_imp['bapq_rigid_total'] = bapq_imp[rigid_cols].sum(axis=1)
bapq_imp['bapq_rigid_ave'] = bapq_imp[rigid_cols].mean(axis=1)

bapq_imp['total_bapq'] = bapq_imp[all_bapq_cols].sum(axis=1)
```

## BIS: Barratt Impulsiveness Scale

```python
bis_imp = pd.read_csv('../phenotype/imputed/imputed_bis01.csv')
```

```python
bis_imp.columns
```

```python
bis_imp['bis_total'] = bis_imp.drop(columns=['subjectkey']).sum(axis=1)
```

```python
# Second-Order Subscales
attentional_impulsiveness = ['bis5', 'bis6', 'bis9', 'bis11', 'bis20', 'bis24', 'bis26', 'bis28']
motor_impulsiveness = ['bis2', 'bis3', 'bis4', 'bis16', 'bis17', 'bis19', 'bis21', 'bis22', 'bis23', 'bis25', 'bis30']
non_planning_impulsiveness = ['bis1', 'bis7', 'bis8', 'bis10', 'bis12', 'bis13', 'bis14', 'bis15', 'bis18', 'bis27', 'bis29']

# First-Order Subscales
attention = ['bis5', 'bis9', 'bis11', 'bis20', 'bis28']
motor = ['bis2', 'bis3', 'bis4', 'bis17', 'bis19', 'bis22', 'bis25']
self_control = ['bis1', 'bis7', 'bis8', 'bis12', 'bis13', 'bis14']
cognitive_complexity = ['bis10', 'bis15', 'bis18', 'bis27', 'bis29']
perseverance = ['bis16', 'bis21', 'bis23', 'bis30']
cognitive_instability = ['bis6', 'bis24', 'bis26']
```

```python
bis_imp['bis_attention_total'] = bis_imp[attention].sum(axis=1)
bis_imp['bis_motor_total'] = bis_imp[motor].sum(axis=1)
bis_imp['bis_self_control_total'] = bis_imp[self_control].sum(axis=1)
bis_imp['bis_cog_complexity_total'] = bis_imp[cognitive_complexity].sum(axis=1)
bis_imp['bis_perseverance_total'] = bis_imp[perseverance].sum(axis=1)
bis_imp['bis_cog_instability_total'] = bis_imp[cognitive_instability].sum(axis=1)
bis_imp['bis_ii_attentional_total'] = bis_imp[attentional_impulsiveness].sum(axis=1)
bis_imp['bis_ii_motor_total'] = bis_imp[motor_impulsiveness].sum(axis=1)
bis_imp['bis_ii_non_planning_total'] = bis_imp[non_planning_impulsiveness].sum(axis=1)
```

```python
bis_imp.head(2)
```

```python
bis_imp.to_csv("../phenotype/imputed/sum_subscale/tot_imputed_bis01.csv", index=False)
```
